```python
import fitz  # PyMuPDF
from PIL import Image
import io
from IPython.display import display

def display_pdf_section_as_image(pdf_path, page_num, crop_area_coords):
    doc = fitz.open(pdf_path)
    if page_num >= doc.page_count:
        print(f"Error: Page number {page_num} out of range.")
        return

    page = doc[page_num]
    crop_rect = fitz.Rect(crop_area_coords)
    zoom = 2
    matrix = fitz.Matrix(zoom, zoom)
    pix = page.get_pixmap(matrix=matrix, clip=crop_rect)
    img = Image.open(io.BytesIO(pix.tobytes()))
    display(img)  # Adjust width and height as needed
    doc.close()

display_pdf_section_as_image(
    pdf_path="C:\\Users\\nicor\\OneDrive\\Documents\\Code\\technojuggling\\sample.pdf", 
    page_num=45,  # First page
    crop_area_coords=(70, 120, 530, 400) # Example coords: from (50,50) to (400,300)
)
```
