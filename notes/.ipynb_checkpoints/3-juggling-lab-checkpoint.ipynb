```python
import requests
from bs4 import BeautifulSoup
import json
import re

def scrape_tumblr_post(post_url):
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36'
    }
    
    response = requests.get(post_url, headers=headers)
    soup = BeautifulSoup(response.text, 'html.parser')
    
    # Tumblr embeds data in JSON-LD script tag
    script_tag = soup.find('script', {'type': 'application/ld+json'})
    
    if script_tag:
        data = json.loads(script_tag.string)
        
        metadata = {
            'blog_name': data.get('author', {}).get('name', ''),
            'date': data.get('datePublished', ''),
            'caption': data.get('headline', ''),
            'description': data.get('description', ''),
            'media_url': data.get('image', {}).get('url', '') if isinstance(data.get('image'), dict) else data.get('image', '')
        }
        
        return metadata
    
    # Fallback: try to find Open Graph meta tags
    metadata = {
        'blog_name': soup.find('meta', {'property': 'og:site_name'})['content'] if soup.find('meta', {'property': 'og:site_name'}) else '',
        'caption': soup.find('meta', {'property': 'og:title'})['content'] if soup.find('meta', {'property': 'og:title'}) else '',
        'description': soup.find('meta', {'property': 'og:description'})['content'] if soup.find('meta', {'property': 'og:description'}) else '',
        'media_url': soup.find('meta', {'property': 'og:image'})['content'] if soup.find('meta', {'property': 'og:image'}) else ''
    }
    
    return metadata
```

```python
import pyperclip

def num_to_height(n):
    if n < 10:
        return str(n)
    elif n <= 35:  # a=10 ... z=35
        return chr(ord('a') + (n - 10))
    else:
        raise ValueError(f"Height {n} exceeds JugglingLab max (z=35)")

def try2(starting=14, ending=3, delay_1=20, delay_2=2):
    def switch(a, b):
        if b == ending:
            return ''
        type_a  = num_to_height(a)
        type_a2 = num_to_height(a + 1)
        type_b  = num_to_height(b)
        type_b2 = num_to_height(b - 1)
        return (
            f'(<{type_a}|{type_b}>^{delay_1})'
            f'<{type_b}p/{type_a}|{type_a}p/{type_a2}p>'
            f'{switch(a + 1, b - 1)}'
            f'(<{type_a2}|{type_b2}>^{delay_2})'
            f'<{type_b2}p/{type_b}p|{type_b}{type_a}p>'
        )

    type_a  = num_to_height(starting)
    a_plus  = num_to_height(starting + 1)
    a_minus = num_to_height(starting - 1)

    output = (
        f'(<{type_a}|{type_a}>^{delay_1})'
        f'<{type_a}{type_a}|{type_a}{a_plus}p>'
        f'{switch(starting + 1, starting - 1)}'
        f'(<{a_plus}|{a_minus}>^{delay_2})'
        f'<{a_minus}p/{type_a}|{type_a}{type_a}>'
    )
    pyperclip.copy(output)
    print(output)

try2(starting=5, ending=1)
```

```python
import fitz  # PyMuPDF
from IPython.display import IFrame

def extract_pdf_pages(input_path, output_path, start_page, end_page):
    """
    Extract specific pages from PDF (0-indexed)
    """
    doc = fitz.open(input_path)
    new_doc = fitz.open()
    
    # Copy pages (0-indexed, so page 1 = index 0)
    for page_num in range(start_page - 1, end_page):
        new_doc.insert_pdf(doc, from_page=page_num, to_page=page_num)
    
    new_doc.save(output_path)
    doc.close()
    new_doc.close()

# Extract pages 5-10
extract_pdf_pages(
    "assets/choreo/live-coding-choreography-real-time-programming-with-terpsicode-sicchio-2023.pdf",
    "assets/choreo/extracted_pages.pdf",
    start_page=2,  # Start at page 5
    end_page=12    # End at page 10
)

# Display extracted PDF
IFrame("assets/choreo/extracted_pages.pdf", width=800, height=600)
```
