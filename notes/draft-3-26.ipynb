### Juggler and Automaton as a Communication Pair
“Juggling seems to appeal to mathematically inclined people - Many amateur jugglers are computer programmers, majors in mathematics or physics, and the like.  One knowledgeable juggler, a mathematics professor himself, told a reporter that forty per cent of jugglers are ‘’algorithmically inclined.” - Claude Shannon

Lesser known about Claude Shannon is his lifelong hobby with juggling and unicycle riding.  While his longstanding accomplishments in mathematics and digital communications are revered, less is acknowledged regarding his mathematical theories of juggling and construction of toy juggling machines.  To further examine the relationship between juggler-technologist and juggling machine, it is useful to consider his Shannon-Weaver model of communication.

This model structures communication between a pair of senders and receivers along a given channel.  In the case of digital communications, it is necessary for an encoding scheme to re-represent analog phenomena like speech, into a digital form which may be transmitted along noisy channels and reconstructed at the receiver's output.  With the presence of feedback, a 2-way communication line is established between the sender and receiver.

This model can be employed to consider relationships between juggler-technologists and a juggling machine or automaton .  Broadening the definition, an automaton may include both mechanical juggling machines like Claude Shannon’s very own WC Fields, and Stephen Meschke’s Pantograph, as well as computer graphical visual simulations of juggling such as Adam Powers, the Amiga Juggler, Juggling Lab, and Marvel’s Spiderman PS4 juggling scene.  While the two mediums of the mechanical and virtual involve vastly different technical challenges, tricks, and language, their interaction potentials exist along a shared continuum.

Even solo juggling itself can be considered through the Shannon-Weaver model as a feedback communication relation between the juggler and juggling props.  The juggler in trying to perform a ball pattern (siteswap), their hands must communicate through the Newtonian language of forces and gravity to interface with balls.  Failure to perform a juggling trick results in dropped balls, the main form of feedback for jugglers.  Hand-eye relations slowly develop muscle memory to execute patterns without “missed packets”.

For automatons, the Shannon-Weaver model is biased towards the right.  Balls are juggled in an unchanging pattern, with minute human intervention.  The machine kinematics of both WC Fields and the Pantograph oscillate in a fixed cadence and range of motion.  After powering on the motor, an operator only needs deft timing to place each ball one by one into the automatons claws to initiate the juggling pattern.  The illusion of juggling is revealed as clawed hands move in the same way with or without balls.  These automations are typically limited to the execution of one pattern like 3 or five ball cascade.

From a different frame, we can also consider a communication relationship between the juggler-technologist and automaton during the design and engineering stages.  Mirroring the feedback of solo-juggling, there is a similar degree of trial and error exchange in which the juggler-technologist may adjust parameters like motor speed, claw size, ball weight, ramp incline, etc. and experimentally test the automatons ability to perform the desired pattern.  This design relation through feedback converges from a shared exchange to the autonomy of a juggling machine.

For virtual juggling simulators, there are a wider variety of communication exchanges and information languages present.  Computer graphics lends itself to a larger degree of illusion, as animations need not perform under the same physical conditions as material props and mechanical limbs.  The use of motion capture to translate the juggling gestures of Ken Rosenthal into the CGI form of Adam Powers, highlights the communicative possibility for tele-operation from physical to virtual.  This was not rendered in real time and involved a degree of post-production to simulate the projectile motions of 3D geometric props.  As the balls themselves were not tracked, the illusion of juggling needed to be carefully simulated to match the timing and positioning of Adam’s hands.  Again, this communication exchange resembles the juggling automatons in that there is an initial degree of feedback formed between the engineer and technological representation; however, when the animation is complete, it serves as a standalone juggler.  The virtual body of Adam Powers could of course be shared more easily than WC Fields due to its playback presence on broadcast television.

Jugglers specifically were among the group of early home computer users and formed online communities like rec.juggling which supplemented existing sharing of juggling news like Juggler’s World Magazine which often featured community submitted pieces and news.  It was on rec.juggling in the late 1980s that tech-savvy jugglers including Jack Boyce built and shared the first siteswap-based juggling animators.  Here the mathematical juggling pattern notation system of siteswap converged with computer graphics in a new way compared to what had been done with Amiga and Adam Powers.  As siteswap became a common means to share juggling patterns in a pre-video internet, siteswap generators forged an early means of exploring and visualizing new juggling patterns.  Now the computer graphical automaton forms a more centralized two way relationship with jugglers through the common language of siteswap.  Serving as a tutor for jugglers to train on new tricks.

Unlike the mechanics of WC Fields, and the motion capture of Adam Powers, these siteswap generators are completely independent of a working knowledge of juggling.  Hand positions and ball paths may be procedurally generated without need for a direct physical juggling source.  Of course it was due to the work of early juggler technologists to translate their own working knowledge of juggling into a computer friendly form.  In siteswap programs, the intermediary of the keyboard and program GUI served as a translation mechanism often moving back and forth between the Newtonian language of bodies and gravitational objects and its computer syntax.  However, three decades after the emergence of siteswap generators, their accessibility improved and actually went viral on contemporary social media platforms most notably on Tumblr.

On Tumblr’s ‘#Juggling Lab’ community, there are hundreds of juggling lab outputs which often diverge from the original intent of sharing siteswaps and into the realm of impossibility and jokes.  It is through this imaginary generated by Juggling Lab users in 2025 that we may further untangle communication relations between users and virtual bodies at the site of keyboard interfaces.

### Juggling Lab

We can probe further into this relation of users with virtual bodies through an examination of Jack Boyce’s software Juggling Lab and its prosumer community on Tumblr.  As this digital community’s popularity peaked in fall of 2025, the program’s original use case of animating siteswaps widened as non-jugglers playfully exploited the interface’s boundaries to create memes.  Juggling Lab through its diverse means of production, unbounded parameter space, image props, and mathematical syntax stimulate users while establishing a freely associative graphics environment.

Before moving forward, it can be helpful to consider Geoff Cox and Alex McLean's description of 'jokes as diagrams of innovative action that represent how humans can diverge from social norms.'  To show how jokes permutate software development, they introduce esoteric programming languages like brainfuck and bodyfuck as two esolangs which intentionally deviate from syntactical norms of traditional programming languages.  While brainfuck is intentionally torturous in its punctuation based notation system, body fuck permutates this by forcing users to type such programs through body gesture recognized computer vision instead of the keyboard.


This is a permutation of language through alteration of its syntax.  In the case of bodyfuck, it even moves the language away from the keyboard and into the body itself.  Different for Juggling Lab though is that its multiparameter siteswap generator renders a 3D graphics scene composed of the dominant computer vision tracked body skeleton anatomy.  A visual animation output instead of the printed console of brainfuck.  This visual output is determined by a numerical notation system for movement patterns that choreographs multiple stick figure bodies, generating significations through image props.  In this way it reverses the user interaction in bodyfuck, where users perform body gestures instead of typing to construct programs.   Surprisingly, a program created initially to aid in the discovery of new juggling patterns finds itself repurposed into an exploratory medium for virtual body expressions.  The Juggling Lab Tumblr community shows how the siteswap and vector based body choreography language lends itself not only to the realm of humor, but also in the formation of a recursive public; A shared social imaginary where alternatives and modifications are made using the Juggling Lab program and community.

While there are some who provide siteswap or program code, there is a general treatment of the program as a sandbox for experimental play.  Jack Boyce of course provides instructions for notational syntax rules regarding stepwise body and hand movements, and common ball patterns.  Many users get started and mess around enough until they land on a funny way to make a joke.  It is most common to use the GUI interface to create jokes since the URL based server does not include image props.  The Juggling Markup Language (JML) is more suited for users familiar with coding in HTML or XML syntax languages and can allow for greater programmability or working organization.

Due to its inclusion of JML, there is a small subset of Juggling Lab users that use other languages like Python to generate JML scripts themselves.  This for example allowed for the rendering of the music videos Bad Apple and Never Gonna Give You Up in Juggling Lab through the downscaling of video resolution where each ball represented a single pixel unit.  Youtube user “IParts970” was first to introduce this approach by overlaying 432 figures each juggling 2 balls, one black and one white, in a 24x18 ball aspect ratio.  Included in their video description is a dropbox link with over 6000 frames from the original Bad Apple video, the jml scripts, and JML output images.  There is no presence of additional program files used to generate, but it is likely impossible that they hand-coded 6000+ files each with 432 jugglers and specification of ball pixels as black or white.  Likely they ran a downscaling resolution script to take the original video and represent it as a 24x18 binary array, and used this to populate a skeleton JML file per frame.  Unlike juggling lab animations, each frame is static and the images had to be stitched together into a single video file.  Likely they needed another script to open up and save the jpg image for each JML instance for it to be recombined as a video file.  Of course the Rick Roll Juggling Lab gif, made by Tumblr user Foone, likely followed IParts970s method given that they used the same 24x18 structure, but instead of a binary black and white, they would need to include a much wider range of ball prop colors.
