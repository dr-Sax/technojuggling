```python
This chapter explores available archives of juggling animation from the first CGI Adam powers, to its occurrence on the Amiga
Commadore, through to a variety of juggling siteswap generator platforms in the 1990s.  Attention will be paid to the technical procedures
involved in creating these animations, the online communities that emerged to support its development, and the humor / self reference present within a 
community developed by STEM academics.

In the late 1970s and 1980s, advancements in 3d software techniques were headed by computer graphics companies who created works for commercial
advertisements and film.  Information International Inc. (III)  is one of those companies comprised of an elite team notably Gary Demos, John Whitney Jr., 
Art Durlinski. etc.  At this time, the demo real was a marketing tool, and III used the demo format as a means to push the envelope in CGI.
They not only rendered the first human figure (adam), but also made it in the form of a juggler.  They used early motion capture techniques by,
taking video feed of a real juggler (ken rosenthal) and mapping the human movements into an animated juggler.  While this demo speaks to a larger
trajectory in computer graphics leading to more sophisticated and realistic rendering of the world into animated form, it also reveals an impulse
in utilizing the juggler as an archetypal form in showcasing new platforms of digital media.  Afterall, Adam Powers was first demo'd at Siggraph
conference in 1981, and is remarked as one of the most memorable and significant events in the evolution of CGI.  Within the imaginary of Adam Powers,
is this conquering of the physical.  Adam represented the real possibiity that one day we could just command a cgi character to juggle 12 balls,
and it would do so on the spot with no training time.  Adam didnt achieve this, but was left as a continued pillar for digital juggling communities.

Adam was followed up by another interesting cgi juggler on the Commadore Amiga platform created by Eric Graham in 1987.  While probably less technically
intensive as Adam Powers, the Amiga Juggler was still a huge marketing tool for the commadore computer platform.  Appearing also in a music video.  Yet 
Amiga is less human like and more like a bubble avatar.  Adam powers was an isolated demo reel, but amiga was an actual piece of software which was
shared around and formed online community of people making variations on the amiga platform.

We can also, see Pixar's take on juggler's through their piece Red's dream depicting the story of a unicycle coming to life in a bicycle shop after hours.
Like toy story, 

As computers are becoming more ubiquitous, we see that the juggling community continous development of juggling avatars.  But it is not by big budget
film companies, rather it is jugglers with positions in STEM departments at universities who are playing around with creating juggling simulators based
on siteswap notation.  At this time, the number of softwares dedicated to juggling simulation is quite vast and difficult to account for in its entirety
It is here that these academic juggling enthusiasts really take up the challenge put forth by III's adam powers in merging a notational syntax of siteswap
with generative juggling patterns bordering on the realm of the unphysical and magic.
```
